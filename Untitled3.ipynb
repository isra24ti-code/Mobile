```python
# ==============================================================================
# Tugas Praktikum Statistika Deskriptif & Penyajian Data
# Nama  : Isra Rahma Wina
# NIM   : 2455301084
# Kelas : 3 TIE
# ==============================================================================

import pandas as pd
import matplotlib.pyplot as plt

# 1. BACA FILE EXCEL
file_path = "IsraRahmaWina3TIE_P2.xlsx"
df_raw = pd.read_excel(file_path, header=7)

# Mengambil data 31 mahasiswa
dataraw = df_raw.iloc[0:31, 1:8].copy()
dataraw.columns = ['NIM', 'Name', 'KUIS', 'UTS', 'UAS', 'Nilai', 'Grade']

print("=== TABEL DATA MAHASISWA ===")
print(dataraw)

# 2. BUAT TABEL FREKUENSI DENGAN CROSSTAB
datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frekuensi")
print("\n=== TABEL FREKUENSI GRADE ===")
print(datafrq)

# 3. SETTING GAYA GRAFIK (PALET UNGU PASTEL)
plt.style.use('seaborn-v0_8-whitegrid')
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle('PENYAJIAN DATA DISTRIBUSI GRADE MAHASISWA', fontsize=14, fontweight='bold', color='#4B0082', y=1.02)

# --- A. GRAFIK GARIS ---
axes[0].plot(datafrq.index, datafrq['Frekuensi'], marker='o', color='#8A2BE2', linewidth=2.5, markersize=8)
axes[0].set_title("Grafik Garis Distribusi Grade", color='#4B0082', fontweight='bold')
axes[0].set_xlabel("Grade", color='#4B0082')
axes[0].set_ylabel("Frekuensi", color='#4B0082')
axes[0].grid(True, linestyle='--', alpha=0.5, color='#D8BFD8')

# --- B. GRAFIK BATANG ---
purple_bars = ['#9370DB', '#BA55D3', '#DDA0DD', '#E6E6FA', '#8A2BE2', '#4B0082']
bars = axes[1].bar(datafrq.index, datafrq['Frekuensi'], color='#B19FF9', edgecolor='#6A5ACD', linewidth=1.5)
axes[1].set_title("Grafik Batang Distribusi Grade", color='#4B0082', fontweight='bold')
axes[1].set_xlabel("Grade", color='#4B0082')
axes[1].set_ylabel("Frekuensi", color='#4B0082')
axes[1].grid(axis='y', linestyle='--', alpha=0.5, color='#D8BFD8')

# Menambahkan angka di atas batang
for bar in bars:
    yval = bar.get_height()
    if yval > 0:
        axes[1].text(bar.get_x() + bar.get_width()/2, yval + 0.2, int(yval), ha='center', va='bottom', color='#4B0082', fontweight='bold')

# --- C. PIE CHART ---
purple_palette = ['#6A0DAE', '#800080', '#9932CC', '#BA55D3', '#DDA0DD', '#E6E6FA']
axes[2].pie(datafrq['Frekuensi'], labels=datafrq.index, autopct='%1.0f%%', colors=purple_palette,
            startangle=140, wedgeprops={'edgecolor': 'white', 'linewidth': 2})
axes[2].set_title("Pie Chart Persentase Grade", color='#4B0082', fontweight='bold')

plt.tight_layout()
plt.show()

# 4. HITUNG STATISTIKA DESKRIPTIF
dataraw["Nilai"] = pd.to_numeric(dataraw["Nilai"], errors='coerce')
dt = dataraw["Nilai"]

stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['variance'] = dt.var()
stats['range'] = dt.max() - dt.min()
stats['skewness'] = dt.skew()
stats['kurtosis'] = dt.kurtosis()

print("\n=== HASIL STATISTIKA DESKRIPTIF ===")
print(stats)
```
